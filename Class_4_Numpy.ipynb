### Task 1

```python
# importing numpy module
import numpy as np

# create 1D Array 5 to 50 jump of 5
array1 = np.arange(5,51,5)
print(type(array1))
print(array1)
```

```python
# reshape the array to 2 rows format
row2_array = array1.reshape(2,5)
print(row2_array)
```

```python
print(row2_array.ndim)
print(row2_array.dtype)
print(row2_array.size)
print(row2_array.shape)
```

### Task 2


```python
marks = np.array([45,82,67,91,38,74,59,88])
```

```python
# How many students scored 60 and above
sixty_plus = marks[marks >= 60]
print(sixty_plus)
```

```python
# Average of 60 plus scores
print(sixty_plus.mean())
```

```python
# Print scores greater than 60 and less than 85
print(sixty_plus[sixty_plus <= 85])
```
